### HANDLING WITH THE OUTLIERS

```python
pd.set_option('display.max_columns', None) #Check all the columns

df[df['pressure_mb']>1500] #Serious outliers
```

```python
cols = [
    "temperature_celsius",
    "pressure_mb",
    "precip_mm",
    "wind_kph",
    "visibility_km",
    "air_quality_PM2.5",
    "air_quality_PM10"
]

print(df[cols].describe().T)
```

```python
from sklearn.preprocessing import StandardScaler

num_cols = ["temperature_celsius", "humidity", "pressure_mb", "wind_kph", "precip_mm",
            "cloud", "visibility_km", "uv_index", "air_quality_PM2.5", "air_quality_PM10"]

# keep df in real units for plots; use df_scaled for distance/scale-sensitive methods
df_scaled = df.copy()
df_scaled[num_cols] = StandardScaler().fit_transform(df[num_cols])
df_scaled[num_cols].describe().T[["mean", "std"]]
```

```python
checks = {
    "temperature_celsius": (df["temperature_celsius"] < -90) | (df["temperature_celsius"] > 60),
    "pressure_mb": (df["pressure_mb"] > 1100),
    "wind_kph": df["wind_kph"] > 200,
    "air_quality_PM2.5": df["air_quality_PM2.5"] > 50,
    "air_quality_PM10": df["air_quality_PM10"] > 50,
}

for col, mask in checks.items():
    print(f"\n{col}: {mask.sum()} suspicious rows")
    print(df.loc[mask, [col]].head(10))
```

##### So there is small amount of suspicious rows

```python
suspicious = (
    (df["temperature_celsius"] > 60) |
    (df["pressure_mb"] < 900) |
    (df["pressure_mb"] > 1100) |
    (df["wind_kph"] > 200) |
    (df["air_quality_PM10"] < 0)
)

df.loc[suspicious, [
    "country",
    "location_name",
    "temperature_celsius",
    "pressure_mb",
    "wind_kph",
    "air_quality_PM10"
]]
```

##### Imputing the outliers (Not removing as the rows are not clearly corrupted)

```python
# physically impossible values -> NaN (real-but-extreme readings stay)
df.loc[df["temperature_celsius"] > 60, "temperature_celsius"] = np.nan
df.loc[df["wind_kph"] > 200, "wind_kph"] = np.nan
df.loc[(df["pressure_mb"] < 900) | (df["pressure_mb"] > 1100), "pressure_mb"] = np.nan
for col in ["air_quality_PM2.5", "air_quality_PM10"]:
    df.loc[df[col] < 0, col] = np.nan

# fill with each city's own median
fix_cols = ["temperature_celsius", "pressure_mb", "wind_kph",
            "air_quality_PM2.5", "air_quality_PM10"]
for col in fix_cols:
    df[col] = df.groupby("location_name")[col].transform(lambda x: x.fillna(x.median()))

print(df[fix_cols].isnull().sum())
```

### ANOMALY DETECTION

```python
from sklearn.ensemble import IsolationForest

# 1) per-city z-score on temperature
g = df.groupby("location_name")["temperature_celsius"]
df["temp_z"] = (df["temperature_celsius"] - g.transform("mean")) / g.transform("std")
df["z_anomaly"] = df["temp_z"].abs() > 3

# 2) multivariate: Isolation Forest
iso_cols = ["temperature_celsius", "humidity", "pressure_mb", "wind_kph",
            "precip_mm", "cloud", "visibility_km", "uv_index"]
iso = IsolationForest(contamination=0.01, random_state=42)
df["iso_anomaly"] = iso.fit_predict(df[iso_cols]) == -1

print("z-score anomalies:", df["z_anomaly"].sum())
print("isolation forest anomalies:", df["iso_anomaly"].sum())
print(df[df["iso_anomaly"]].groupby("country").size().sort_values(ascending=False).head(10))
```

```python
city_df["temp_lag_1"] = city_df["temperature_celsius"].shift(1)
city_df["temp_lag_7"] = city_df["temperature_celsius"].shift(7)
city_df["temp_lag_14"] = city_df["temperature_celsius"].shift(14)

city_df["temp_roll_7"] = (
    city_df["temperature_celsius"]
    .rolling(7)
    .mean()
)

city_df["temp_roll_30"] = (
    city_df["temperature_celsius"]
    .rolling(30)
    .mean()
)

city_df["month"] = city_df["last_updated"].dt.month
city_df["day_of_year"] = city_df["last_updated"].dt.dayofyear

city_df = city_df.dropna()
```

```python
# what do the anomalies look like in Tokyo?
tk = df[df["location_name"] == "Tokyo"].sort_values("last_updated")
plt.figure(figsize=(14, 5))
plt.plot(tk["last_updated"], tk["temperature_celsius"], lw=1, label="Temperature")
hit = tk[tk["z_anomaly"]]
plt.scatter(hit["last_updated"], hit["temperature_celsius"], color="red", zorder=3, label="Anomaly (|z|>3)")
plt.title("Tokyo temperature anomalies")
plt.legend()
plt.show()
```

```python
monthly = df.groupby(["month", "region"])["temperature_celsius"].mean().unstack()
monthly.plot(marker="o", figsize=(12, 6))
plt.title("Average temperature by month and region")
plt.ylabel("°C")
plt.show()

df.groupby("region")["precip_mm"].mean().sort_values().plot.barh(figsize=(8, 5))
plt.title("Average precipitation by region (mm)")
plt.show()
```

```python
aq_cols = [c for c in df.columns if c.startswith("air_quality")]
weather_cols = ["temperature_celsius", "humidity", "pressure_mb", "wind_kph",
                "precip_mm", "cloud", "visibility_km", "uv_index"]

aq_corr = df[aq_cols + weather_cols].corr(method="spearman").loc[aq_cols, weather_cols]
plt.figure(figsize=(11, 6))
sns.heatmap(aq_corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Air quality vs weather (Spearman)")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18, 6))
plot_data = (
    df.groupby(["location_name", "latitude", "longitude"], as_index=False)
      .agg(
          temp=("temperature_celsius", "mean"),
          pm25=("air_quality_PM2.5", "mean")
      )
)

for ax, col, title in zip(
    axes,
    ["temp", "pm25"],
    ["Mean temperature (°C)", "Mean PM2.5"]
):
    sc = ax.scatter(
        plot_data["longitude"],
        plot_data["latitude"],
        c=plot_data[col],
        cmap="coolwarm",
        s=25,
        alpha=0.7
    )
    fig.colorbar(sc, ax=ax, label=col)
    ax.set_title(title)
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
plt.show()

# top/bottom countries
country_pm = df.groupby("country")["air_quality_PM2.5"].mean().sort_values()
pd.concat([country_pm.head(5), country_pm.tail(5)]).plot.barh(figsize=(8, 5))
plt.title("Cleanest vs most polluted countries (mean PM2.5)")
plt.show()
```
